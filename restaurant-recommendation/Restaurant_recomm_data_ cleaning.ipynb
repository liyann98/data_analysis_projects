Merging cust and loc datasets to form con dataset.

Removing the extra lower case redundant data from gender and location_type

```python
def gender(x):
    try:
        x1=x.split(" ")[0]
        x1=x1.lower()
        return x1
    except:
        return np.nan
con.gender=con.gender.apply(gender)
```

```python
def loc(x):
    try:
        x1=x.split(" ")[0]
        x1=x1.lower()
        return x1
    except:
        return np.nan
con.location_type=con.location_type.apply(loc)
```

Merging con and odr datasets to create the final dataset.

```python
df=pd.merge(con,odr,on='customer_id')
df_final=pd.merge(df,ven, on='vendor_id')
```

Converting prep_time to integer.

```python
prep_time=[]
def convert(x):
    global prep_time
    if x not in prep_time:
        prep_time.append(x)
    return prep_time.index(x)
df_final['prepration_time']=df_final['prepration_time'].apply(convert)
```
