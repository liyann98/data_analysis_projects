```python
import os
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
# define the directory path to the Tests folder
tests_dir = "./Tests"

# create empty list to store all the dataframes for each test
df_list = []

# loop through all the folders in the Tests directory
for folder in os.listdir(tests_dir):
    
    # if the folder is not a directory, skip it
    if not os.path.isdir(os.path.join(tests_dir, folder)):
        continue
        
    # get the path to the Bilgi.txt file for the current folder
    bilgi_file = os.path.join(tests_dir, folder, "Bilgi.txt")
    
    # read in the data from the Bilgi.txt file
    with open(bilgi_file, "r") as f:
        data = f.readlines()
    
    # extract the gender, age, height, and weight from the data
    gender = data[0].strip().split(":")[1].strip()
    weight = data[1].strip().split(":")[1].strip()
    height = data[2].strip().split(":")[1].strip()
    age = data[3].strip().split(":")[1].strip()
    
    # loop through all the activity folders for the current volunteer
    for activity in os.listdir(os.path.join(tests_dir, folder, "Testler Export")):
        
        # if the activity is not a directory, skip it
        if not os.path.isdir(os.path.join(tests_dir, folder, "Testler Export", activity)):
            continue
        
        # loop through all the test folders for the current activity
        for test_folder in os.listdir(os.path.join(tests_dir, folder, "Testler Export", activity)):
            
            # if the test folder is not a directory, skip it
            if not os.path.isdir(os.path.join(tests_dir, folder, "Testler Export", activity, test_folder)):
                continue
            
            # loop through all the sensor files for the current test
            for sensor_file in os.listdir(os.path.join(tests_dir, folder, "Testler Export", activity, test_folder)):
                
                # if the file is not a txt file, skip it
                if not sensor_file.endswith(".txt"):
                    continue
                
                # get the path to the current sensor file
                sensor_path = os.path.join(tests_dir, folder, "Testler Export", activity, test_folder, sensor_file)
                
                # read in the data from the sensor file
                sensor_data = pd.read_csv(sensor_path, sep="\t",error_bad_lines=False,skiprows=5, names=["Counter","Temperature","VelInc_X",
                                                                          "VelInc_Y","VelInc_Z","OriInc_w","OriInc_x","OriInc_y",
                                                                          "OriInc_z","Acc_X","Acc_Y","Acc_Z","Gyr_X","Gyr_Y","Gyr_Z",
                                                                          "Mag_X","Mag_Y","Mag_Z","Pressure","Roll","Pitch","Yaw","RSSI"])
                
                # add columns for gender, age, height, and weight
                sensor_data["Gender"] = gender
                sensor_data["Age"] = age
                sensor_data["Height"] = height
                sensor_data["Weight"] = weight
                
                # add columns for volunteer, activity, and test number and Sensor
                sensor_data["Volunteer"] = folder
                sensor_data["Activity"] = activity
                sensor_data["Test"] = test_folder[-1]
                
                # add a column for sensor name based on the text file name
                if sensor_file == "340506.txt":
                    sensor_data["Sensor"] = "Head Sensor"
                elif sensor_file == "340527.txt":
                    sensor_data["Sensor"] = "Chest Sensor"
                elif sensor_file == "340535.txt":
                    sensor_data["Sensor"] = "Waist Sensor"
                elif sensor_file == "340537.txt":
                    sensor_data["Sensor"] = "Right Wrist Sensor"
                elif sensor_file == "340539.txt":
                    sensor_data["Sensor"] = "Right Thigh Sensor"
                elif sensor_file == "340540.txt":
                    sensor_data["Sensor"] = "Right Ankle Sensor"
                else:
                    sensor_name = sensor_file.split('.')[0]
                    sensor_data["Sensor"] = f"Unknown Sensor ({sensor_name})"
                   
                # add the sensor data for the current test to the df_list
                df_list.append(sensor_data)
                
# concatenate all the dataframes in the df_list into a
df = pd.concat(df_list, ignore_index=True)
```
