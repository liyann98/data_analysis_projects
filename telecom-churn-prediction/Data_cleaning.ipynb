#### Dropping the Churn Category and the Churn Reason column as there are too many missing values

```python
df = df.drop(['Churn Category', 'Churn Reason'], axis=1)
```

#### Taking the mean of the columns to fill in numeric missing values

```python
df['Avg Monthly Long Distance Charges'] = df['Avg Monthly Long Distance Charges'].fillna(df['Avg Monthly Long Distance Charges'].mean())
```

```python
df['Avg Monthly GB Download']=df['Avg Monthly GB Download'].fillna(df['Avg Monthly GB Download'].mean())
```

#### Using the mode of the columns to fill in the missing categorical columns

```python
df['Multiple Lines']= df['Multiple Lines'].fillna(df['Multiple Lines'].mode().iloc[0])
df['Internet Type']= df['Internet Type'].fillna(df['Internet Type'].mode().iloc[0])
df['Online Security']= df['Online Security'].fillna(df['Online Security'].mode().iloc[0])
df['Online Backup']= df['Online Backup'].fillna(df['Online Backup'].mode().iloc[0])
df['Device Protection Plan']= df['Device Protection Plan'].fillna(df['Device Protection Plan'].mode().iloc[0])
df['Premium Tech Support']= df['Premium Tech Support'].fillna(df['Premium Tech Support'].mode().iloc[0])
df['Streaming TV']= df['Streaming TV'].fillna(df['Streaming TV'].mode().iloc[0])
df['Streaming Movies']= df['Streaming Movies'].fillna(df['Streaming Movies'].mode().iloc[0])
df['Streaming Music']= df['Streaming Music'].fillna(df['Streaming Music'].mode().iloc[0])
df['Unlimited Data']= df['Unlimited Data'].fillna(df['Unlimited Data'].mode().iloc[0])
```

#### Changing the dtypes from object to numeric and categorical respectively.

```python
#Converting columns to numeric
numeric_columns = ['Age', 'Number of Dependents', 'Zip Code', 'Latitude', 'Longitude', 'Number of Referrals', 'Tenure in Months',
                   'Avg Monthly Long Distance Charges', 'Avg Monthly GB Download', 'Monthly Charge', 'Total Charges',
                   'Total Refunds', 'Total Extra Data Charges', 'Total Long Distance Charges', 'Total Revenue']

df[numeric_columns] = df[numeric_columns].apply(pd.to_numeric, errors='coerce')
```

```python
#Converting columns to category
category_columns = ['Gender', 'Married', 'City', 'Offer', 'Phone Service', 'Multiple Lines', 'Internet Service',
                    'Internet Type', 'Online Security', 'Online Backup', 'Device Protection Plan',
                    'Premium Tech Support', 'Streaming TV', 'Streaming Movies', 'Streaming Music',
                    'Unlimited Data', 'Contract', 'Paperless Billing', 'Payment Method', 'Customer Status']

df[category_columns] = df[category_columns].astype('category')
```
